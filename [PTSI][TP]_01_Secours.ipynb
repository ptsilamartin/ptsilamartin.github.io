# Les premières boucles

## La boucle `for`
La boucle `for` est une structure algorithmique qui permet de répéter une série d'instructions un certain nombre de fois. Les instructions qui seront répétées sont indentées de 4 espaces. 

En Python, la fonction `range` permet de renvoyer une séquence de nombres entiers. L'appel `range(5)` permet de retourner les nombres `(0,1,2,3,4)`.


```python
# Exemple
def affichage_nombre(n:int) -> None:
    """
    Affichage des nombres de 0 à n-1.
    """
    for i in range(n):
        print(i)   
```

```python
affichage_nombre(5)
```

## La boucle while

La boucle `while` est une structure algorithmique qui permet de répéter une série d'instructions tant qu'une condition booléenne est vraie. Les instructions qui seront répétées sont indentées de 4 espaces.

```python
# Exemple
def affichage_nombre_while(n:int) -> None:
    """
    Affichage des nombres de 0 à n-1.
    """
    i = 0
    while i < n:
        print(i)
        i = i+1
```

```python
affichage_nombre_while(5)
```
